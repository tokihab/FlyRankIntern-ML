## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
import pandas as pd
import os

# Load dataset
if os.path.exists("../../data/raw/content_refresh_anonymized.csv"):
    df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
else:
    df = pd.read_csv("https://raw.githubusercontent.com/tokihab/flyrankintern-ml/main/data/raw/content_refresh_anonymized.csv")

# 1. Build the feature vector (handling missing values)
features = ["word_count", "avg_position", "competition", "search_volume"]
X = df[features].copy()
X["word_count"] = X["word_count"].fillna(0) # Impute missing word counts with 0
print("Feature vector built successfully. Shape:", X.shape)
display(X.head())
```

## 2. Feature notes (meaning, missing, categorical, available-when?)
- **word_count**: Length of page content; missing values (31%) imputed with 0; available before decision time.
- **avg_position**: Average Google rank; measures visibility; available historically before decision time.
- **competition**: Market competition score for queries; available at decision time.
- **search_volume**: Monthly search demand; available prior to refresh choice.

```python
# 3. The leakage hunt test: ensure no future metrics or target-derived leakage exist
leakage_candidate = "views_after"
if leakage_candidate in df.columns:
    print(f"WARNING: Found {leakage_candidate} - checking correlation with target.")
    # In our audit, views_after showed 99% importance and was dropped.
else:
    print("Leakage check passed: 'views_after' and future metrics are successfully excluded.")
```
