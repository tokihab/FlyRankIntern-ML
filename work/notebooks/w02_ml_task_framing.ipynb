## 5. Why ML beats a fixed rule here

What makes the pattern too messy for an if-statement?

A human writing SQL IF/THEN statements can only easily cross two or three dimensions (e.g., word_count < 1000 AND impressions > 500).
A clustering algorithm like K-Means or DBSCAN evaluates 10+ dimensions simultaneously (traffic, engagement, text length, age, position) to find natural multi-dimensional groupings that a human could never hardcode by hand.
